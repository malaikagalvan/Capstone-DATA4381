### Main code for manual sorting

```python
# Names of the directories
label_dir = ""
signal_dir = ""
background_dir = ""

# Ensure the correct directories are created.
os.makedirs(signal_dir, exist_ok=True); 
os.makedirs(background_dir, exist_ok=True);

# Creates a backup directory.
label_dir_copy = label_dir + "_copy"
if not os.path.exists(label_dir_copy):
    shutil.copytree(label_dir, label_dir_copy)
    print("Backup folder created.")

# Makes sure that no copies exist in the labeled folders.
original_folder = sorted(os.listdir(label_dir_copy))
files=[]
for f in original_folder:
    signal = os.path.exists(os.path.join(signal_dir, f))
    background = os.path.exists(os.path.join(background_dir, f))
    if not (signal or background):
        files.append(f)

# Function to display images.
def show_image(img):
    plt.figure(figsize=(4,4))
    plt.imshow(cv2.cvtColor(img, cv2.COLOR_BGR2RGB))
    plt.axis("off")
    plt.show()



# Main loop
rand.shuffle(files) # made looping random to reduce rater bias (in case of retries).
for index, filename in enumerate(files, start=1):
    path = os.path.join(label_dir_copy, filename)
    img = cv2.imread(path)
    if img is None:
        continue
        
    clear_output(wait=True)
    progress_pct = (index / len(files)) * 100
    print(f"Image {index}/{len(files)}: {filename}   {progress_pct:.2f}% done." )
    show_image(img)

    key = input("Label (1=signal, 2=background, q=quit): ").strip().lower()

    if key == "1": 
        cv2.imwrite(os.path.join(signal_dir, filename), img)
        if os.path.exists(path):
            os.remove(path)
    elif key == "2": 
        cv2.imwrite(os.path.join(background_dir, filename), img)
        if os.path.exists(path):
            os.remove(path)
    elif key == "q":
        print("Exiting labeling.")
        break   
```
